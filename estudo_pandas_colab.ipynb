```python
def is_dataframe_grande(df: pd.DataFrame):
  total_de_colunas = df.shape[1]
  print(df.shape)
  print(f'O total de colunas do dataframe é: {total_de_colunas}')

  if total_de_colunas > 15:
    print(f'Dataframe grande')
  else:
    print(f'Dataframe pequeno')

df.head(5)

is_dataframe_grande(df)
```

```python
def size_linhas(df: pd.DataFrame):
  total_de_linhas = df.shape[0]
  print(f'O total de linhas do dataframe é: {total_de_linhas}')

  if total_de_linhas > 100:
    print(f'Dataframe grande')
  else:
    print(f'Dataframe pequeno')

size_linhas(df)

size_linhas(df.head(5))
```
